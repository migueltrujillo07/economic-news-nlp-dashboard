```python
# ---
# PROJECT: Economic News NLP Dashboard
# MODULE: Descriptive Analysis (EDA)
# AUTHOR: Miguel Trujillo
# ---

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import os
import nltk
from nltk.corpus import stopwords

# 1. SETTINGS & LOAD DATA
# Setting aesthetics for professional plots
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Load the fixed filename from the preprocessing stage
DATA_PATH = "../data/processed/clean_economic_news_v1.csv"

if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset successfully loaded. Total records: {len(df)}")
else:
    print("Error: The processed file was not found. Please run src/preprocess.py first.")

# 2. DATA QUALITY CHECK
# Checking for null values and data types to ensure integrity
print("\n--- Data Integrity Check ---")
print(df.info())
print("\nMissing values per column:")
print(df.isnull().sum())


# 3. TEXT STRUCTURE ANALYSIS
# Calculating word count to understand the density of the news titles
df['word_count'] = df['clean_title'].apply(lambda x: len(str(x).split()))


# 4. N-GRAM ANALYSIS (PHRASE FREQUENCY)
# Identifying the most frequent 2-word combinations (Bigrams) to find economic topics
def get_top_bigrams(corpus, n=10):
    words = " ".join(corpus.astype(str)).split()
    # Basic filtering: common Spanish connectors (Stopwords)
    nltk.download('stopwords')
    stop_words = set(stopwords.words('english'))
    filtered_words = [w for w in words if w not in stop_words]
    
    bigrams = zip(filtered_words, filtered_words[1:])
    counts = Counter(bigrams)
    return counts.most_common(n)

print("\n--- Top 10 Economic Phrases (Bigrams) ---")
top_bigrams = get_top_bigrams(df['clean_title'])
for gram, freq in top_bigrams:
    print(f"{' '.join(gram)}: {freq}")

print("\n--- Analyst Insights ---")
print(f"1. The dataset is concentrated in {len(df['source_name'].unique())} different sources.")
print(f"2. Average title length is {df['word_count'].mean():.1f} words.")
print(f"3. Most frequent topic detected: '{' '.join(top_bigrams[0][0]) if top_bigrams else 'N/A'}'")
```
